# Data Quality: Distribution Report
This interactive notebook loads the staged generic dataset and automatically visualizes its null distributions and categorical variables.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# 1. Load the staged data
df = pd.read_csv('../../staged/staged_users.csv')
print(f"Data Loaded: {df.shape[0]} rows, {df.shape[1]} columns")
display(df.head())
```

## 1. Null Value Analysis
The following table identifies columns containing missing values.

```python
null_counts = df.isnull().sum()
null_pct = (null_counts / len(df)) * 100

null_summary = pd.DataFrame({
    'Null Count': null_counts,
    'Null Percentage': null_pct
}).sort_values(by='Null Percentage', ascending=False)

display(null_summary[null_summary['Null Count'] > 0])
```

## 2. Categorical Variable Distributions
The bar charts below illustrate the exact frequency counts of distinct values for all categorical (string/object/category) features in the dataset. Missing values are explicitly represented as *Missing (NaN)*.

```python
categorical_cols = df.select_dtypes(include=['object', 'category']).columns

for col in categorical_cols:
    plt.figure(figsize=(10, 4))
    
    # Temporarily fill NA so it plots visibly on the chart
    plot_data = df[col].fillna('Missing (NaN)')
    
    # Generate countplot sorted by frequency
    sns.countplot(y=plot_data, order=plot_data.value_counts().index, palette='crest', hue=plot_data, legend=False)
    
    plt.title(f'Distribution of {col}', fontweight='bold')
    plt.xlabel('Frequency Count')
    plt.ylabel(col)
    plt.tight_layout()
    plt.show()
```
